```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import to_date

spark = SparkSession.builder.enableHiveSupport().getOrCreate()
df = spark.read.parquet("/user/hive/warehouse/sclsepb4532_raw.db/clickstream_logs", header=True, inferSchema=True)
```

```python
df = df.withColumn("event_date", to_date("event_timestamp"))
```

```python
df.write \
    .mode("overwrite") \
    .partitionBy("event_date") \
    .option("path", "/user/hive/warehouse/sclsepb4532_staging.db/clickstream_logs_stg") \
    .saveAsTable("sclsepb4532_staging.clickstream_logs_stg")
```
