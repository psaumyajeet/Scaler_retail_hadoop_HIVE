```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col
from pyspark.sql import functions as F

spark = SparkSession.builder.enableHiveSupport().getOrCreate()
df1 = spark.read.parquet("/user/sclsepb4532/sclsepb4532_staging.db/orders_stg", header=True, inferSchema=True)
df2 = spark.read.parquet("/user/sclsepb4532/sclsepb4532_staging.db/customers_stg", header=True, inferSchema=True)
```

```python
df3 = df1.join(df2, df1['customer_id']==df2['customer_id'],'inner')
```
