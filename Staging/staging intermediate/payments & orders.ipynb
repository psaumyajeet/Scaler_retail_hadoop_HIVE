```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.enableHiveSupport().getOrCreate()

df1 = spark.read.parquet("/user/sclsepb4532/sclsepb4532_staging.db/payments_stg", header=True, inferSchema=True)
df2 = spark.read.parquet("/user/sclsepb4532/sclsepb4532_staging.db/orders_stg", header=True, inferSchema=True)
```

```python
df3 = df1.join(df2, df1['payment_id']==df2['payment_id'],'inner')
```

```python
df3 = df3.drop(df2['payment_id'])
```

```python
df3 = df3.drop(df2['order_id'])
```

```python
df3 = df3.drop(df2['total_amount'])
```

```python
df3.write.mode("overwrite").saveAsTable("sclsepb4532_staging.payments_n_order")
```
