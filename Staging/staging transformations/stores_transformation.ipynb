```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, trim, to_date

spark = SparkSession.builder.enableHiveSupport().getOrCreate()
df = spark.read.parquet("/user/hive/warehouse/sclsepb4532_raw.db/stores", header=True, inferSchema=True)
```

```python
stores_stg_df = df.select(
    trim(col("store_id")).alias("store_id"),
    trim(col("store_name")).alias("store_name"),
    trim(col("city")).alias("city"),
    trim(col("state")).alias("state"),
    trim(col("region")).alias("region"),
    to_date(trim(col("opening_date")), "yyyy-MM-dd").alias("opening_date"),    
    trim(col("store_type")).alias("store_type")
)
df = stores_stg_df.filter(col("store_id").isNotNull())
```

```python
df.write \
    .mode("overwrite") \
    .saveAsTable("sclsepb4532_staging.stores_stg")
```
