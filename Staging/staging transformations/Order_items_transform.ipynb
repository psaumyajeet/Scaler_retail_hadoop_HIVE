```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, round

spark = SparkSession.builder.enableHiveSupport().getOrCreate()
df = spark.read.parquet("/user/hive/warehouse/sclsepb4532_raw.db/order_items", header=True, inferSchema=True)
```

```python
# Apply transformations sequentially
df = df \
    .filter(col("quantity") > 0) \
    .filter(col("product_id").isNotNull() & (col("product_id") != "")) \
    .withColumn(
        "item_total", round((col("quantity") * col("item_price")) - col("discount_amount"), 2))
```

```python
df.write \
    .mode("overwrite") \
    .saveAsTable("sclsepb4532_staging.order_items_stg")
```
