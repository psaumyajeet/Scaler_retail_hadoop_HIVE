```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col

spark = SparkSession.builder.getOrCreate()
df = spark.read.csv("/user/sclsepb4532/orders.csvb0445dfb-15b5-4408-8c3c-8039dd01afe2_table", header=True, inferSchema=True)
```

```python
df_cleaned = df.filter("total_amount >= 0")
```

```python
valid_statuses = [
    "Delivered",
    "Shipped",
    "Cancelled",
    "Placed",
]

# Keep only rows with valid order_status values
df_cleaned = df.filter(col("order_status").isin(valid_statuses))
```

```python
df_cleaned = df_cleaned.filter(
    col("order_date").isNotNull() &
    (col("order_date") != "")
)
```
