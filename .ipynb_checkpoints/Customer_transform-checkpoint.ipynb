```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import avg, col

spark = SparkSession.builder.getOrCreate()
df = spark.read.csv("/user/sclsepb4532/customers.csv81abaa5f-8f42-479e-8122-d093f1d4736f_table", header=True, inferSchema=True)
```

```python
# Remove duplicates based on the 'customer_id' column
df_cleaned = df.dropDuplicates(["customer_id"])
```

```python
# Read data
data = df.select("email").show(truncate=False)
df = spark.createDataFrame(data, ["name", "email"])

# Standardize the email column to lowercase
df_cleaned = df.withColumn("email", lower(df["email"]))

# Show the results
```

```python
import pyspark.sql.functions as F

# Overwrite the existing "name" column with trimmed values
#df_cleaned = df.withColumn("name", F.trim(F.col("first_name")))
```

```python
from pyspark.sql import functions as F

# Trim outer whitespace and reduce inner multiple spaces to a single space
df_cleaned = df.withColumn("first_name", F.trim(F.regexp_replace(F.col("first_name"), r"\s+", " "))) \
               .withColumn("last_name", F.trim(F.regexp_replace(F.col("last_name"), r"\s+", " ")))
```

```python
# Replace null values only in the phone_number column
df_cleaned = df.fillna("Unknown", subset=["phone_number"])
df_cleaned.show()
```

```python
valid_types = ["Offline", "Online", "Both"]
df_cleaned = df.withColumn("customer_type",
    F.when(F.col("customer_type").isin(valid_types), F.col("customer_type"))
     .otherwise("Other")) # Catch-all for invalid data or NULLs)

df_cleaned.show()
```
