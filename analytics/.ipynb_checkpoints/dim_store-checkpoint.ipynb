```python
from pyspark.sql import functions as F
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, sha2, when, trim

spark = SparkSession.builder.enableHiveSupport().getOrCreate()
store_stg = spark.read.parquet("/user/sclsepb4532/sclsepb4532_staging.db/stores_stg")
```

```python
dim_store = store_stg.select(
    sha2(col("store_id"), 256).alias("store_key"),
    col("region"),
    col("city"),
        when(
        trim(col("city")).isin("Mumbai", "Delhi", "Bangalore", "Chennai", "Kolkata"), "Tier1")
    .otherwise("Tier2").alias("Store_type"))
```

```python
dim_store.write\
.mode("overwrite")\
.option("compression", "snappy")\
.saveAsTable("sclsepb4532_analytics.dim_store_analytics")
```
